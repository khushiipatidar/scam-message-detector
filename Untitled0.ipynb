```python
import pandas as pd

data = [
 ("spam","aapka KYC expire ho gaya hai abhi link par click karein"),
 ("spam","aapne 10 lakh ki lottery jeeti hai OTP bhejein"),
 ("spam","aapka bank account band ho jayega turant details update karein"),
 ("spam","free recharge paane ke liye is link par click karo"),
 ("spam","aapka ATM card block ho gaya hai abhi call karein"),
 ("spam","congratulations aapko iPhone mila hai claim karne ke liye link kholo"),
 ("spam","aapka electricity bill pending hai aaj raat connection kat jayega"),
 ("spam","ghar baithe din ke 5000 kamao abhi WhatsApp karein"),
 ("spam","PM yojana me aapka naam aaya hai 2000 rupay paane ke liye form bharein"),
 ("spam","aapka PAN card suspend ho gaya hai verify karne ke liye click karein"),
 ("ham","kal 5 baje meeting hai aa jana"),
 ("ham","ghar pahunch gaya chinta mat karna"),
 ("ham","aaj raat khane me kya banega"),
 ("ham","mummy ne kaha sabzi leke aana"),
 ("ham","exam ka time table aa gaya hai check kar lena"),
 ("ham","bhai kal cricket khelne chalenge kya"),
 ("ham","assignment kal tak submit karna hai yaad rakhna"),
 ("ham","happy birthday yaar party kab de raha hai"),
 ("ham","main thodi der me call karta hoon"),
 ("ham","notes bhej dena kal ke lecture ke"),
]
pd.DataFrame(data, columns=["label","text"]).to_csv("hindi.csv", index=False)
print("hindi.csv ban gayi")
```

```python
df1 = pd.read_csv("spam.csv", encoding="latin-1")[["v1", "v2"]]
df1.columns = ["label", "text"]

df2 = pd.read_csv("hindi.csv")

df = pd.concat([df1, df2], ignore_index=True)
df["label"] = df["label"].map({"ham": 0, "spam": 1})
df["text"] = df["text"].str.lower()
print(df["label"].value_counts())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import classification_report, confusion_matrix

X_train, X_test, y_train, y_test = train_test_split(
    df["text"], df["label"], test_size=0.2, random_state=42, stratify=df["label"]
)

model = Pipeline([
    ("tfidf", TfidfVectorizer(ngram_range=(1, 2), max_features=5000)),
    ("clf", MultinomialNB())
])
model.fit(X_train, y_train)

pred = model.predict(X_test)
print(classification_report(y_test, pred))
print(confusion_matrix(y_test, pred))
```

```python
tests = [
    "aapka KYC expire ho gaya hai turant link par click karein",
    "aapko lottery me 5 lakh mile hain OTP share karein",
    "kal college jaana hai kya",
    "congratulations you won a free prize call now",
]
for t in tests:
    p = model.predict_proba([t.lower()])[0][1]
    print(f"{p*100:.1f}% scam -> {t}")
```

```python
extra = [
 ("spam","aapka SBI account freeze ho gaya hai turant KYC update karein link par click karein"),
 ("spam","aapko lottery me 5 lakh rupay mile hain OTP bhejkar claim karein"),
 ("spam","aapka Paytm KYC pending hai warna account band ho jayega"),
 ("spam","aapke number par 25 lakh ka inaam nikla hai abhi call karein"),
 ("spam","aapka Aadhaar link nahi hai abhi verify karein warna SIM block hoga"),
 ("spam","loan paas ho gaya hai bina documents ke 50000 paaye abhi apply karein"),
 ("spam","aapka parcel rok diya gaya hai customs fee bharne ke liye link kholein"),
 ("spam","UPI se 1 rupay bhejein aur 5000 cashback paayein"),
 ("spam","aapki Jio SIM 24 ghante me band ho jayegi abhi KYC karwayein"),
 ("spam","paise double karne ka mauka abhi invest karein guaranteed return"),
 ("ham","aaj class me kya padhaya tha"),
 ("ham","papa station pahunch gaye hain unko lene jaana hai"),
 ("ham","dinner ke liye paneer le aana"),
 ("ham","kal ki chhutti hai ya college hai"),
 ("ham","project ka kaam kahan tak hua"),
 ("ham","tera phone nahi lag raha call kar dena"),
]
old = pd.read_csv("hindi.csv")
new = pd.concat([old, pd.DataFrame(extra, columns=["label","text"])], ignore_index=True)
new.to_csv("hindi.csv", index=False)

df2 = pd.read_csv("hindi.csv")
df = pd.concat([df1, df2], ignore_index=True)
df["label"] = df["label"].map({"ham": 0, "spam": 1})
df["text"] = df["text"].str.lower()

from sklearn.linear_model import LogisticRegression
X_train, X_test, y_train, y_test = train_test_split(
    df["text"], df["label"], test_size=0.2, random_state=42, stratify=df["label"]
)
model = Pipeline([
    ("tfidf", TfidfVectorizer(ngram_range=(1, 2), max_features=5000)),
    ("clf", LogisticRegression(max_iter=1000, class_weight="balanced"))
])
model.fit(X_train, y_train)
pred = model.predict(X_test)
print(classification_report(y_test, pred))
print(confusion_matrix(y_test, pred))
```
